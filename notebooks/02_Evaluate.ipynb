```python
import pandas as pd
import matplotlib.pyplot as plt
actual_df = pd.read_csv("data/processed/weekly_features_weekly_v1.csv")
```

```python
#Forecast
forecast_paths = [
    "outputs/forecasts/2026-09-18/xgboost_group_global_v1_ensemble/forecast.csv", #21/9 to 25/9
    "outputs/forecasts/2026-09-25/xgboost_group_global_v1_ensemble/forecast.csv", #28/9 to 2/10
    #"outputs/forecasts/2026-10-02/xgboost_group_global_v1_ensemble/forecast.csv", #5/10 to 9/10
]
```

```python
#Target (Used Lag1)
target_weeks = [
    "2026-09-25", #21/9 to 25/9
    "2026-10-02", #28/9 to 2/10
    #"2026-10-09",
]
```

```python
import numpy as np
record = {}
for target_week, forecast_path in zip(target_weeks,forecast_paths):
    actual = actual_df[actual_df['target_week'] == target_week][['ticker','target']]
    predict = pd.read_csv(forecast_path).sort_values(by="ticker")[['ticker','predicted_weekly_return']]    
    df_merge = pd.merge(actual, predict)
    df_merge['abs_error'] = np.abs(df_merge.target - df_merge.predicted_weekly_return)
    mae = df_merge['abs_error'].mean()
    record[target_week] = {'df':df_merge, "mae":mae}

```

```python
for key in record:
    record_dict = record[key]
    print('='*10 + f"Friday of forecast week: {key}" + '='*10)
    print(record_dict['df'])
    print(f"MAE :{record_dict['mae']}")
    
```
